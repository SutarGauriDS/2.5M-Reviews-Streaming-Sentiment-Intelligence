```python
from pyspark.sql import functions as F
from pyspark.ml.feature import VectorAssembler, StandardScaler
from pyspark.ml.clustering import KMeans

# ============================================================
# SILVER LAYER - INCOME & SPENDING DATA
# ============================================================

# ------------------------------------------------------------
# 1. TABLE NAMES
# ------------------------------------------------------------

bronze_table = "income_etlpipeline.default.income_bronze"
silver_table = "income_etlpipeline.default.income_silver"


# ------------------------------------------------------------
# 2. READ BRONZE TABLE
# ------------------------------------------------------------

df_silver = spark.table(bronze_table)

print("Bronze data loaded successfully")
print("Bronze Row Count:", df_silver.count())

display(df_silver)


# ------------------------------------------------------------
# 3. CLEAN DATA
# ------------------------------------------------------------

# Remove duplicate records
df_silver = df_silver.dropDuplicates()

# Remove rows containing null values
df_silver = df_silver.dropna()

print("Rows after cleaning:", df_silver.count())


# ------------------------------------------------------------
# 4. CREATE FEATURES FOR K-MEANS
# ------------------------------------------------------------

assembler = VectorAssembler(
    inputCols=["Income", "SpendingScore"],
    outputCol="features"
)

df_silver = assembler.transform(df_silver)


# ------------------------------------------------------------
# 5. SCALE FEATURES
# ------------------------------------------------------------

scaler = StandardScaler(
    inputCol="features",
    outputCol="scaled_features",
    withMean=True,
    withStd=True
)

scaler_model = scaler.fit(df_silver)

df_silver = scaler_model.transform(df_silver)


# ------------------------------------------------------------
# 6. K-MEANS CUSTOMER SEGMENTATION
# ------------------------------------------------------------

kmeans = KMeans(
    k=4,
    seed=42,
    featuresCol="scaled_features",
    predictionCol="KMeans_Cluster"
)

kmeans_model = kmeans.fit(df_silver)

df_silver = kmeans_model.transform(df_silver)


# ------------------------------------------------------------
# 7. VIEW CLUSTERING RESULTS
# ------------------------------------------------------------

df_silver.select(
    "CustomerID",
    "Income",
    "SpendingScore",
    "KMeans_Cluster"
).show(20, truncate=False)


# ------------------------------------------------------------
# 8. VISUALIZATION
# ------------------------------------------------------------

import matplotlib.pyplot as plt

plot_df = (
    df_silver
    .select(
        "Income",
        "SpendingScore",
        "KMeans_Cluster"
    )
    .toPandas()
)

plt.figure(figsize=(10, 6))

plt.scatter(
    plot_df["Income"],
    plot_df["SpendingScore"],
    c=plot_df["KMeans_Cluster"],
    s=50
)

plt.xlabel("Income")
plt.ylabel("Spending Score")
plt.title("K-Means Customer Segmentation")
plt.colorbar(label="K-Means Cluster")

plt.show()


# ------------------------------------------------------------
# 9. SAVE SILVER TABLE
# ------------------------------------------------------------

(
    df_silver
    .write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable(silver_table)
)

print("Silver layer created successfully!")
print("Silver Table:", silver_table)


# ------------------------------------------------------------
# 10. VERIFY SILVER TABLE
# ------------------------------------------------------------

silver_df = spark.table(silver_table)

print("Silver Row Count:", silver_df.count())

display(silver_df)
```
