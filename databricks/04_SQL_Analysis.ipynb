```sql
USE CATALOG streaming_sentiment_intelligence;

USE SCHEMA default;
```

```sql
SHOW TABLES;
```

```sql
SELECT
    COUNT(*) AS total_reviews,
    COUNT(DISTINCT user) AS unique_users,
    COUNT(DISTINCT platform) AS total_platforms,
    COUNT(DISTINCT country) AS total_countries,
    ROUND(AVG(likes), 2) AS avg_likes,
    ROUND(AVG(retweets), 2) AS avg_retweets,
    ROUND(AVG(engagement_score), 2) AS avg_engagement
FROM silver_reviews;
```

```sql
SELECT
    sentiment,
    COUNT(*) AS review_count,
    ROUND(
        COUNT(*) * 100.0 /
        SUM(COUNT(*)) OVER (),
        2
    ) AS sentiment_percentage
FROM silver_reviews
GROUP BY sentiment
ORDER BY review_count DESC;
```

```sql
SELECT
    platform,
    COUNT(*) AS total_reviews,
    ROUND(AVG(likes), 2) AS avg_likes,
    ROUND(AVG(retweets), 2) AS avg_retweets,
    ROUND(AVG(engagement_score), 2) AS avg_engagement
FROM silver_reviews
GROUP BY platform
ORDER BY total_reviews DESC;
```

```sql
SELECT
    platform,
    sentiment,
    COUNT(*) AS review_count,
    ROUND(AVG(engagement_score), 2) AS avg_engagement
FROM silver_reviews
GROUP BY platform, sentiment
ORDER BY platform, review_count DESC;
```

```sql
SELECT
    country,
    COUNT(*) AS total_reviews,
    ROUND(AVG(likes), 2) AS avg_likes,
    ROUND(AVG(retweets), 2) AS avg_retweets,
    ROUND(AVG(engagement_score), 2) AS avg_engagement
FROM silver_reviews
GROUP BY country
ORDER BY total_reviews DESC;
```

```sql
SELECT
    country,
    COUNT(*) AS total_reviews,
    ROUND(AVG(likes), 2) AS avg_likes,
    ROUND(AVG(retweets), 2) AS avg_retweets,
    ROUND(AVG(engagement_score), 2) AS avg_engagement
FROM silver_reviews
GROUP BY country
ORDER BY total_reviews DESC;
```

```sql
SELECT
    review_text,
    sentiment,
    platform,
    country,
    likes,
    retweets,
    engagement_score
FROM silver_reviews
ORDER BY engagement_score DESC
LIMIT 20;
```

```sql
SELECT
    review_text,
    sentiment,
    platform,
    country,
    likes,
    retweets,
    engagement_score
FROM silver_reviews
ORDER BY engagement_score DESC
LIMIT 20;
```

```sql
SELECT
    sentiment,
    COUNT(*) AS review_count,
    ROUND(AVG(text_length), 2) AS avg_text_length,
    ROUND(AVG(word_count), 2) AS avg_word_count,
    ROUND(AVG(engagement_score), 2) AS avg_engagement
FROM silver_reviews
GROUP BY sentiment
ORDER BY avg_engagement DESC;
```

```sql
SELECT
    hour,
    sentiment,
    COUNT(*) AS review_count,
    ROUND(AVG(engagement_score), 2) AS avg_engagement
FROM silver_reviews
GROUP BY hour, sentiment
ORDER BY hour, review_count DESC;
```

```sql
SELECT
    year,
    month,
    sentiment,
    COUNT(*) AS review_count,
    ROUND(AVG(engagement_score), 2) AS avg_engagement
FROM silver_reviews
GROUP BY year, month, sentiment
ORDER BY year, month, review_count DESC;
```

```sql
SELECT
    hashtag,
    COUNT(*) AS usage_count
FROM (
    SELECT
        explode(
            split(
                regexp_replace(hashtags, '#', ''),
                ' '
            )
        ) AS hashtag
    FROM silver_reviews
    WHERE hashtags IS NOT NULL
)
WHERE TRIM(hashtag) <> ''
GROUP BY hashtag
ORDER BY usage_count DESC
LIMIT 20;
```

```sql
SELECT
    platform,
    ROUND(AVG(engagement_score), 2) AS avg_engagement,
    SUM(likes) AS total_likes,
    SUM(retweets) AS total_retweets
FROM silver_reviews
GROUP BY platform
ORDER BY avg_engagement DESC;
```

```sql
SELECT
    platform,
    country,
    COUNT(*) AS negative_reviews,
    ROUND(AVG(engagement_score), 2) AS avg_engagement
FROM silver_reviews
WHERE LOWER(sentiment) IN (
    'negative',
    'anger',
    'sadness',
    'fear'
)
GROUP BY platform, country
ORDER BY negative_reviews DESC;
```

```sql
SELECT
    platform,
    country,
    COUNT(*) AS positive_reviews,
    ROUND(AVG(engagement_score), 2) AS avg_engagement
FROM silver_reviews
WHERE LOWER(sentiment) = 'positive'
GROUP BY platform, country
ORDER BY positive_reviews DESC;
```
